```python
import pandas as pd 
df_gold = pd.read_csv('../data/silver/retail_sales_clean.csv')
df_gold.shape
```

```python
category_sales = df_gold.groupby('category')['total_amount'].sum()
category_sales = category_sales.reset_index()
category_sales = category_sales.rename(
    columns={'total_amount': 'total_revenue'}
)
category_sales = category_sales.sort_values(
    by='total_revenue', ascending=False
)
display(category_sales)
```

```python
city_sales = df_gold.groupby('city')['total_amount'].sum()
city_sales = city_sales.reset_index()
city_sales = city_sales.rename(
    columns={'total_amount': 'total_revenue'}
)
city_sales = city_sales.sort_values(
    by='total_revenue', ascending=False
)
display(city_sales)
```

```python
df_gold['order_date'] = pd.to_datetime(df_gold['order_date'])

monthly_sales = df_gold.assign(
    month=df_gold['order_date'].dt.to_period('M').astype(str)
).groupby('month')['total_amount'].sum()

monthly_sales = monthly_sales.reset_index()
monthly_sales = monthly_sales.rename(
    columns={'total_amount': 'total_revenue'}
)
display(monthly_sales)
```

```python
product_sales = df_gold.groupby('product')['total_amount'].sum()
product_sales = product_sales.reset_index()
product_sales = product_sales.rename(
    columns={'total_amount': 'total_revenue'}
)
product_sales = product_sales.sort_values(
    by='total_revenue', ascending=False
)
display(product_sales)
```

```python
category_sales.to_csv('../data/quality_reports/category_sales.csv', index=False)

city_sales.to_csv('../data/quality_reports/city_sales.csv', index=False)

monthly_sales.to_csv('../data/quality_reports/monthly_sales.csv', index=False)

product_sales.to_csv('../data/quality_reports/product_sales.csv', index=False)
```
