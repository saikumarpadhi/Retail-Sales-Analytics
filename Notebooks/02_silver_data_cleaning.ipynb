```python
import pandas as pd

df = pd.read_csv('../data/bronze/retail_sales.csv')
df.head()
```

```python
df.isnull().sum().sum()
```

```python
df[df.duplicated()]
```

```python
df_clean = df.drop_duplicates()
df_clean.shape
```

```python
df_clean.duplicated().sum()
```

```python
df_clean['customer_name'] = df_clean['customer_name'].fillna('Unknown')
df_clean["customer_name"].isnull().sum()
```

```python
df_clean['age'].median()
```

```python
df_clean['age'] = df_clean['age'].fillna(df_clean['age'].median())
df_clean['age'].isnull().sum()
```

```python
df_clean['city'].isnull().sum()
```

```python
df_clean['city'] = df_clean['city'].fillna('Unknown')
df_clean['city'].isnull().sum()
```

```python
df_clean['payment_method'].isnull().sum()
```

```python
df_clean['payment_method'] = df_clean['payment_method'].fillna('Unknown')
df_clean['payment_method'].isnull().sum()
```

```python
df_clean["discount"].isnull().sum()
```

```python
df_clean[df_clean["discount"].isnull()].head(10)
```

```python
missing_discounts = df_clean[df_clean['discount'].isnull()].copy()
missing_discounts['calc_discounts'] = (
     1 - missing_discounts['total_amount']/(missing_discounts[
    'quantity'] * missing_discounts['unit_price'])
    ).round(2)
```

```python
missing_discounts[["quantity", "unit_price", "total_amount", "calc_discounts"]].head(10)
```

```python
missing_discounts["calc_discounts"].round(2).value_counts()
```

```python
df_clean.loc[missing_discounts.index,'discount'] = (missing_discounts["calc_discounts"].round(2))
df_clean["discount"].isnull().sum()
```

```python
df_clean["discount"].value_counts().sort_index()
```

```python
df_clean.isnull().sum()
```

```python
df_clean["city"].value_counts()
```

```python
df_clean['city'] = df_clean['city'].str.strip()
```

```python
df_clean['city'] = df_clean['city'].str.title()
```

```python
df_clean['category'] = df_clean['category'].str.strip().str.title()
```

```python
df_clean["payment_method"] = df_clean['payment_method'].str.strip().str.title()
```

```python
df_clean["payment_method"] = df_clean["payment_method"].replace("Upi", "UPI")
```

```python
df_clean["order_date"].dtype
```

```python
df_clean["order_date"] = pd.to_datetime(df_clean["order_date"])
```

```python
df_clean["order_date"].dt.year.value_counts().sort_index()
```

```python
df_clean["order_date"].dt.month.value_counts().sort_index()
```

```python
df_clean.shape
```

```python
df_clean.to_csv("../data/silver/retail_sales_clean.csv", index=False)
```
